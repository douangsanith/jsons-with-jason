### 1 · Lists — a matter docket

A **list** is an ordered collection, written with square brackets `[ ]`. Think of it as a **docket**: several matters lined up in order. You reach an item by its **position number**, and Python counts from **0**.

````python
docket = ["M-1001", "M-1002", "M-1003", "M-1004"]

print(docket[0])        # first item  -> M-1001  (counting starts at 0!)
print(docket[1])        # second item -> M-1002
print(docket[-1])       # last item   -> M-1004
print(len(docket))      # how many    -> 4

docket.append("M-1005") # add one to the end
print(docket)
````

**What just happened:** the docket is a single variable holding many matter numbers. `[0]` is the first, `len()` counts them, `.append()` adds one.

### 2 · Dictionaries — one matter as a labeled record

A list is great for *a bunch of the same thing*, but a matter has *different labeled fields* — client, area, amount. That's a **dictionary**: `key: value` pairs in curly braces `{ }`. Analogy: a **completed intake form**, where each blank has a label.

````python
matter = {
    "matter_id": "M-1002",
    "client": "Brightline LLC",
    "practice_area": "Litigation",
    "amount_billed": 42750.50,
    "status": "Active",
}

print(matter["client"])         # look up a field by its label
print(matter["amount_billed"])

matter["lead_attorney"] = "S. Patel"  # add a new field
print(matter)
````

**What just happened:** we described *one* matter with named fields and looked them up by name (`matter["client"]`) instead of by position.

### 3 · The real shape of data: a list of dictionaries

Put them together and you get the shape of nearly all tabular data: **a list (the docket) of dictionaries (each matter)**. This is exactly what a spreadsheet row-per-matter looks like — and what we'll load into pandas on Day 4.

````python
docket = [
    {"matter_id": "M-1001", "client": "Acme Corp",     "amount_billed": 18500.00, "status": "Active"},
    {"matter_id": "M-1002", "client": "Brightline LLC", "amount_billed": 42750.50, "status": "Active"},
    {"matter_id": "M-1005", "client": "Evergreen Inc",  "amount_billed":  2450.00, "status": "On Hold"},
]

print("Matters in docket:", len(docket))
print("First client:", docket[0]["client"])   # first matter, then its client field
````

### 4 · For loops — review every matter automatically

A **for loop** says *“for each item in the collection, do this.”* Instead of writing the same line five times, you write it once and Python repeats it. This is the moment code starts saving you real time.

````python
for matter in docket:
    # this block runs once per matter; `matter` is the current dictionary
    print(f'{matter["matter_id"]}: {matter["client"]} — ${matter["amount_billed"]:,.2f}')
````

**What just happened:** the loop handed us each matter in turn and we printed a summary line for it. Add a 4th matter to `docket` above and it just works — no extra code.

### 5 · Conditionals — make a decision

**if / elif / else** lets code branch: *if the amount is high, flag it; otherwise, don't.* The `==`, `>`, `<` comparisons return `True`/`False` (remember `bool` from Day 1).

````python
HIGH_VALUE = 25000   # our threshold for "needs partner review"

for matter in docket:
    amount = matter["amount_billed"]
    if amount >= HIGH_VALUE:
        flag = "🔴 HIGH VALUE"
    elif matter["status"] == "On Hold":
        flag = "⏸️  on hold"
    else:
        flag = "✅ routine"
    print(f'{matter["matter_id"]}: {flag}')
````

> **`Go Deeper 🔧` — list comprehensions & `.get()`.** A comprehension builds a new list in one line. And `dict.get("key", default)` safely returns a fallback instead of crashing when a field is missing.

````python
# Go Deeper (optional)
# 1) one-line list of just the high-value client names
big = [m["client"] for m in docket if m["amount_billed"] >= 25000]
print(big)

# 2) .get() avoids a crash when a key might be missing
print(docket[0].get("lead_attorney", "(unassigned)"))
````

> **`Common pitfalls ⚠️`**
>
> - Positions start at **0**, so the 3rd item is `docket[2]`.
> - Looking up a missing key with `matter["foo"]` raises **KeyError** — use `.get()` if unsure.
> - Use `==` to compare (`status == "Active"`), not `=`.

### 🚀 Build the artifact — a docket triage report

Combine list + dict + loop + conditional into a small **triage report** that flags what needs attention. This is a genuine mini-tool — point it at any docket and it reviews every matter for you.

````python
docket = [
    {"matter_id": "M-1001", "client": "Acme Corp",     "amount_billed": 18500.00, "status": "Active"},
    {"matter_id": "M-1006", "client": "Foster & Sons",  "amount_billed": 58900.75, "status": "Active"},
    {"matter_id": "M-1005", "client": "Evergreen Inc",  "amount_billed":  2450.00, "status": "On Hold"},
    {"matter_id": "M-1011", "client": "Keystone Bank",  "amount_billed": 71200.00, "status": "On Hold"},
]

HIGH_VALUE = 25000
print("MATTER TRIAGE REPORT")
print("-" * 40)
attention = 0
for m in docket:
    notes = []
    if m["amount_billed"] >= HIGH_VALUE:
        notes.append("high-value")
    if m["status"] == "On Hold":
        notes.append("on hold")
    label = ", ".join(notes) if notes else "routine"
    if notes:
        attention += 1
    print(f'{m["matter_id"]}  {m["client"]:<16} {label}')
print("-" * 40)
print(f"{attention} of {len(docket)} matters need attention.")
print("\n✅ Shipped: an automated docket triage report.")
````

- **for loops** repeat work over every item; **if/elif/else** makes decisions.
