### ⚙️ Setup

The `show()` helper plus a few reusable **matter cards** to arrange. No installs.

````python
from IPython.display import HTML, display

def show(html: str):
    """Render an HTML string inline, exactly as a browser would."""
    display(HTML(html))

# three little cards we'll lay out in different ways
CARDS = """
  <div class="card"><h3>M-1002</h3><p>Brightline LLC</p><p class="amt">$42,750.50</p></div>
  <div class="card"><h3>M-1001</h3><p>Acme Corp</p><p class="amt">$18,500.00</p></div>
  <div class="card"><h3>M-1004</h3><p>Dovetail Inc</p><p class="amt">$9,800.00</p></div>
"""

CARD_CSS = """
<style>
  .wrap { font-family: -apple-system, Segoe UI, Roboto, sans-serif; }
  .wrap .card { border: 1px solid #e5e7eb; border-radius: 10px; padding: 14px; background: #f9fafb; min-width: 140px; }
  .wrap .card h3 { margin: 0 0 6px 0; color: #1e3a8a; }
  .wrap .card p { margin: 2px 0; color: #374151; font-size: 14px; }
  .wrap .card .amt { font-weight: 600; }
</style>
"""

print("Setup ready — CARDS and CARD_CSS defined.")
````

### 1 · Normal flow — the default

Before any layout, HTML stacks in **normal flow**: **block** elements (`div`, `p`, `h1`, `section`) each take a full line, top to bottom; **inline** elements (`a`, `span`, `strong`) sit within a line. Watch three cards with no layout — they stack.

````python
show(CARD_CSS + '<div class="wrap">' + CARDS + '</div>')
````

**What just happened:** the cards are `<div>`s (block), so they stack vertically — the browser's default. Layout tools change this.

### 2 · Flexbox — lay out a **row**

Put `display: flex` on a **container** and its children line up in a **row**. The key properties:
- **`gap`** — space between items.
- **`justify-content`** — spacing along the row (`flex-start`, `space-between`, `center`).
- **`align-items`** — vertical alignment (`stretch`, `center`).

> **Legal analogy:** flex is arranging exhibits **side by side on a table** and evenly spacing them.

````python
show("""
<style>
  .row { display: flex; gap: 12px; justify-content: space-between; }
</style>
""" + CARD_CSS + '<div class="wrap"><div class="row">' + CARDS + '</div></div>')
````

**What just happened:** `display: flex` turned the stack into a **row**; `gap` spaced the cards; `justify-content: space-between` pushed them to the edges with even gaps. Change `justify-content` to `center` or `flex-start` to see them regroup.

### 3 · Responsive — let the row **wrap**

On a narrow screen three fixed cards overflow. **`flex-wrap: wrap`** lets items drop to the next line when they don't fit, and `flex: 1` lets each card grow to share the space. That's "responsive" with two words.

````python
show("""
<style>
  .rrow { display: flex; gap: 12px; flex-wrap: wrap; }
  .rrow .card { flex: 1 1 160px; }   /* grow, shrink, ideal width 160px */
</style>
""" + CARD_CSS + '<div class="wrap"><div class="rrow">' + CARDS + '</div></div>')
````

**What just happened:** `flex-wrap: wrap` + `flex: 1 1 160px` means "aim for ~160px each, grow to fill, and wrap to a new line when there's no room." Resize the browser (on the saved file) and the cards re-flow — no fixed pixel math.

### 4 · CSS grid — a **dashboard grid**

For rows *and* columns, **grid** is cleaner. `display: grid` + `grid-template-columns` defines the columns; `gap` spaces them. `repeat(auto-fit, minmax(160px, 1fr))` is the famous "responsive grid in one line."

````python
show("""
<style>
  .grid { display: grid; grid-template-columns: repeat(auto-fit, minmax(160px, 1fr)); gap: 12px; }
</style>
""" + CARD_CSS + '<div class="wrap"><div class="grid">' + CARDS + CARDS + '</div></div>')
````

**What just happened:** the grid auto-fits as many ~160px columns as fit, each sharing leftover space (`1fr`), wrapping to new rows automatically. Six cards flow into a tidy grid — the shape of most dashboards.

> **`Go Deeper 🔧` — flex vs grid, and reading it.** **Flex** is best for **one dimension** (a row *or* a column — toolbars, card rows). **Grid** is best for **two dimensions** (rows *and* columns — page/dashboard layouts). When you open a Claude artifact, search the CSS for `display: flex` / `display: grid` — that instantly tells you how the page is arranged and where to change it.

> **`Common pitfalls ⚠️`**
>
> - **Flex/grid properties go on the _container_,** not the items (`display: flex` on the parent).
> - **`justify-content` vs `align-items`:** justify = main axis (across the row); align = cross axis (up/down).
> - **Forgetting `flex-wrap: wrap`** is why a row overflows on small screens.
> - **`gap` replaces margins** between flex/grid items — cleaner than per-item margins.

### 🚀 Build the artifact — a responsive matters dashboard

````python
dashboard = ("""
<style>
  .board { display: grid; grid-template-columns: repeat(auto-fit, minmax(160px, 1fr)); gap: 12px; }
</style>
""" + CARD_CSS + '<div class="wrap"><h2 style="font-family:sans-serif;color:#111827;">Active matters</h2>'
   + '<div class="board">' + CARDS + CARDS + '</div></div>')

document = "<!DOCTYPE html>\n<html>\n<head>\n<title>Matters dashboard</title>\n</head>\n<body>\n" + dashboard + "\n</body>\n</html>\n"
with open("matters_dashboard.html", "w") as f:
    f.write(document)
print("📄 Saved matters_dashboard.html — open it and resize the window to watch it re-flow.")
show(dashboard)
````

### 📖 Reference & glossary

| Term | Plain meaning |
|---|---|
| **Normal flow** | default stacking (block vs inline) |
| **Flexbox** | 1-D layout on a container: `display: flex` |
| **`justify-content`** | spacing along the main axis |
| **`align-items`** | alignment on the cross axis |
| **`gap`** | space between flex/grid items |
| **`flex-wrap: wrap`** | let items wrap to new lines |
| **Grid** | 2-D layout: `display: grid` |
| **`grid-template-columns`** | define the columns |
| **`repeat(auto-fit, minmax(...))`** | responsive columns in one line |
