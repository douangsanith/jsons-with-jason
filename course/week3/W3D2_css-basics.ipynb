### ⚙️ Setup

### 1 · What CSS is, and how to apply it

**CSS** (Cascading Style Sheets) is a list of **rules**: *"for these elements, set these properties."* A rule is a **selector** plus a `{ property: value; }` block. Three places it can live:
- **Inline** — a `style="…"` attribute on one element (quick, but doesn't scale).
- **`<style>` block** — a set of rules in the page (what Claude usually does).
- **External `.css` file** — shared across pages (the production norms; `Go Deeper`).

````python
# inline style — the quickest form
show('<h1 style="color: teal;">Matter Intelligence</h1>')

# a <style> block — a rule that targets ALL <h2> elements
show("""
<style>
  h2 { color: #b45309; }     /* every h2 turns amber */
</style>
<h2>Top active matters</h2>
<h2>Notes</h2>
""")
````

**What just happened:** the inline `style` colored one heading; the `<style>` rule colored *every* `<h2>`. A rule reads **selector `{ property: value; }`** — here selector `h2`, property `color`, value `#b45309` (a hex color). `/* … */` is a CSS comment.

### 2 · Selectors — tag, `.class`, `#id`

How you target elements:
- **`h2`** — every `<h2>` tag.
- **`.matters`** — every element with `class="matters"` (a **dot** = class).
- **`#footer`** — the element with `id="footer"` (a **hash** = id; ids are unique).

Classes are the workhorse — reusable across many elements.

````python
show("""
<style>
  .subtitle  { color: gray; font-style: italic; }   /* class selector */
  .disclaimer { color: #b91c1c; font-size: 12px; }
</style>
""" + artifact)
````

**What just happened:** we prepended a `<style>` block to the artifact and targeted its hooks — `.subtitle` and `.disclaimer` — leaving the HTML untouched. **This is the core move:** change the look by targeting classes, not by editing structure.

### 3 · The box model — space with `padding`, `border`, `margin`

Every element is a **box** with three layers of spacing around its content:
- **`padding`** — space *inside* the box, between content and border.
- **`border`** — the line around the box.
- **`margin`** — space *outside* the box, pushing other elements away.

> **Legal analogy:** the box model is a document's **margins and indents** — padding is the whitespace inside a text box, margin is the gap to the next block.

````python
show("""
<style>
  #demo-card {
    padding: 16px;
    border: 1px solid #d1d5db;
    border-radius: 8px;      /* rounded corners */
    margin: 12px;
    background: #f9fafb;
  }
</style>
<div id="demo-card">
  <strong>M-1002 — Brightline LLC</strong><br>
  Litigation · $42,750.50
</div>
""")
````

**What just happened:** the `#demo-card` rule turned a plain `<div>` into a **card** — padding gave it breathing room, a border + `border-radius` framed it, `background` tinted it, `margin` set it off. Cards like this are the building block of every dashboard.

### 4 · Put it together — style the whole artifact

Now target the artifact's hooks in one `<style>` block: a header bar, a clean table, a subtle card around the whole thing. Everything is scoped under **`.mi-dashboard`** so these styles don't leak to the rest of the notebook.

````python
styled = """
<style>
  .mi-dashboard {
    font-family: -apple-system, Segoe UI, Roboto, sans-serif;
    max-width: 640px; margin: 12px 0; padding: 20px;
    border: 1px solid #e5e7eb; border-radius: 12px; color: #111827;
  }
  .mi-dashboard h1 { color: #1e3a8a; margin: 0 0 4px 0; font-size: 22px; }
  .mi-dashboard .subtitle { color: #6b7280; margin-top: 0; }
  .mi-dashboard table.matters { border-collapse: collapse; width: 100%; }
  .mi-dashboard table.matters th,
  .mi-dashboard table.matters td { text-align: left; padding: 8px 10px; border-bottom: 1px solid #eef2f7; }
  .mi-dashboard table.matters th { background: #f3f4f6; font-size: 13px; color: #374151; }
  .mi-dashboard .disclaimer { color: #9ca3af; font-size: 12px; }
</style>
""" + artifact

show(styled)
````

**What just happened:** no HTML changed — only CSS — yet the plain page now reads like a real dashboard. That separation (structure fixed, style swappable) is exactly why you can re-skin a Claude artifact to your brand quickly.

> **`Go Deeper 🔧` — external stylesheets & specificity.**
> - Production apps put CSS in a separate file: `<link rel="stylesheet" href="styles.css">`. Same rules, shared across pages.
> - When two rules target the same element, **specificity** decides: `#id` beats `.class` beats `tag`; an inline `style="…"` beats them all. If a style "won't apply," a more specific rule is usually overriding it.

> **`Common pitfalls ⚠️`**
>
> - **Notebook style leak:** a `<style>` in one cell can affect other cells. **Scope** rules under a container class (`.mi-dashboard …`) — good practice everywhere, essential here.
> - **`.class` vs `#id`:** dot for class, hash for id. Mixing them up is the #1 "why isn't it styling?" bug.
> - **Units:** most sizes need a unit — `padding: 16px`, not `padding: 16`.
> - **Every declaration ends with `;`** inside the `{ }` block.
