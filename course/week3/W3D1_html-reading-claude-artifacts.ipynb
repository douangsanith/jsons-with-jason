### ⚙️ Setup

No installs. We use `IPython.display.HTML`, which renders an HTML string **right in the notebook** — so you see the page as a browser would. `show(html)` is our tiny helper.

````python
from IPython.display import HTML, display

def show(html: str):
    """Render an HTML string inline, exactly as a browser would."""
    display(HTML(html))

show("<p>✅ Setup done. If you can read this sentence, HTML is rendering.</p>")
````

### 1 · HTML is made of **elements** (tags)

An **HTML element** is a piece of content wrapped in **tags**: an opening tag `<p>`, the content, and a closing tag `</p>`. `p` means "paragraph." The tag name in angle brackets tells the browser *what kind of thing* this is.

> **Legal analogy:** a tag is a **labeled field on a form**. `<p>…</p>` says "this part is a paragraph," the way a form labels a box "Client Name." The label isn't shown to the reader — it tells the system what the content *is*.

````python
show("<h1>Matter Intelligence</h1>")     # h1 = the biggest heading
show("<p>A dashboard for our active matters.</p>")   # p  = a paragraph
````

**What just happened:** `<h1>` rendered a large heading, `<p>` a normal paragraph — same words, different *tags*, so the browser treats them differently. Headings run `<h1>` (largest) down to `<h6>` (smallest).

### 2 · Elements **nest** inside each other

Elements go *inside* other elements, forming a hierarchy. A `<ul>` (unordered list) contains `<li>` (list item) elements. Indentation isn't required, but everyone uses it to show the nesting.

> **Legal analogy:** nesting is **clauses inside articles inside a contract** — a list of items lives inside the section that lives inside the document.

````python
show("""
<ul>
  <li>M-1001 — Acme Corp</li>
  <li>M-1002 — Brightline LLC</li>
  <li>M-1004 — Dovetail Inc</li>
</ul>
""")
````

**What just happened:** one `<ul>` wrapped three `<li>` items — a bulleted list. The `<li>`s are **children** of the `<ul>`. (Triple-quoted `"""…"""` just lets us write multi-line HTML in Python.)

### 3 · Tags can carry **attributes**

An **attribute** is extra information inside the opening tag, written `name="value"`. The most important ones:
- **`href`** on a link `<a>` — *where it goes*.
- **`class`** and **`id`** — labels used to *style* or *find* an element (you'll use these constantly next lessons).

> **Legal analogy:** attributes are the **parameters on a defined term** — the same clause, tuned by its inputs (effective date, jurisdiction). `<a href="…">` is a cross-reference with its target filled in.

````python
show('<a href="https://example.com/docket">Open the full docket</a>')
show('<p class="disclaimer" id="footer-note">Synthetic data — not legal advice.</p>')
````

**What just happened:** the `<a>` became a clickable link because of its `href` attribute; the `<p>` carries a `class` and an `id` (invisible now, but hooks we'll use for styling in Day 2). **Attribute values go in quotes.**

### 4 · The tags Claude uses to **structure a page**

Claude doesn't dump raw text — it uses **semantic tags** that name each region of the page:

| Tag | Meaning | Brief analogy |
|---|---|---|
| `<header>` | top banner (title, intro) | the caption/heading of a brief |
| `<main>` | the primary content | the body |
| `<section>` | a grouped chunk of content | an article/section |
| `<table>` | tabular data (rows & columns) | an exhibit table |
| `<ul>` / `<li>` | a bulleted list & its items | a list of points |
| `<footer>` | bottom notes | the signature block / disclaimer |

Here's a small **Claude-style artifact** built from those tags. Read it, then run it.

````python
artifact = """
<header>
  <h1>Matter Intelligence</h1>
  <p>Active matters overview — Legal Innovation Team.</p>
</header>

<main>
  <section>
    <h2>Top active matters</h2>
    <table>
      <thead>
        <tr><th>Matter</th><th>Client</th><th>Practice area</th><th>Billed</th></tr>
      </thead>
      <tbody>
        <tr><td>M-1002</td><td>Brightline LLC</td><td>Litigation</td><td>$42,750.50</td></tr>
        <tr><td>M-1001</td><td>Acme Corp</td><td>Contracts</td><td>$18,500.00</td></tr>
        <tr><td>M-1004</td><td>Dovetail Inc</td><td>Employment</td><td>$9,800.00</td></tr>
      </tbody>
    </table>
  </section>

  <section>
    <h2>Notes</h2>
    <ul>
      <li>Data is synthetic — no real client information.</li>
      <li>See the <a href="https://example.com/docket">full docket</a> for details.</li>
    </ul>
  </section>
</main>

<footer>
  <p>Not legal advice · a lawyer reviews any AI output.</p>
</footer>
"""

show(artifact)
````

**What just happened:** every visible piece — the title, the table of matters, the notes list, the footer disclaimer — is one of those semantic tags. This is exactly the shape of a real Claude artifact (minus the styling, which is tomorrow). A `<table>` is built from **rows** `<tr>`, with header cells `<th>` and data cells `<td>`.

### 5 · A page is a **tree** (the DOM)

Because elements nest, an HTML page is a **tree** — each element is a branch, its children hang beneath it. Browsers call this the **DOM** (Document Object Model). Seeing the tree is how you navigate any artifact.

The artifact above, as a tree:

```
header
├─ h1            "Matter Intelligence"
└─ p             "Active matters overview…"
main
├─ section
│  ├─ h2         "Top active matters"
│  └─ table
│     ├─ thead → tr → th × 4
│     └─ tbody → tr × 3 → td × 4
└─ section
   ├─ h2         "Notes"
   └─ ul → li × 2   (one li contains an <a> link)
footer
└─ p             "Not legal advice…"
```

> **Legal analogy:** the DOM is a **document outline** — articles containing sections containing clauses. To change one clause you find its place in the outline; to change one label you find its element in the tree.

### 6 · Reading a Claude artifact — the 3 questions

When Claude hands you HTML, read it by asking:
1. **What are the regions?** Scan for `<header>` / `<main>` / `<section>` / `<footer>` — that's the outline.
2. **Where's the data?** Find the `<table>`, `<ul>`, or repeated blocks — that's what you'll swap for real matters.
3. **What's a hook?** Note `id`/`class`/`href` attributes — those are what you (or a developer) grab to style or wire up.

That's the entire skill: **outline → data → hooks.** Everything else this week builds on it.

> **`Go Deeper 🔧` — the full document skeleton.** What we rendered is a *fragment*. A complete artifact Claude saves as a `.html` file wraps it in a document skeleton:
>
> ```html
> <!DOCTYPE html>
> <html>
>   <head>
>     <title>Matter Intelligence</title>   <!-- tab title; not shown on the page -->
>     <style> /* CSS lives here — Day 2 */ </style>
>   </head>
>   <body>
>     …the header/main/footer you just read…
>   </body>
> </html>
> ```
>
> `<head>` holds *metadata* (title, styles) the visitor doesn't see directly; `<body>` holds everything on the page. We render fragments inline because the notebook is already a web page — but you'll see this full skeleton when you open a saved artifact. (Its anatomy is Day 4.)

> **`Common pitfalls ⚠️`**
>
> - **Every tag needs a closing tag** (`<p>…</p>`). Forget the `</p>` and the rest of the page can misbehave. *(A few "void" tags like `<br>` and `<img>` have no closer — the exception.)*
> - **Attribute values need quotes:** `href="…"`, not `href=…`.
> - **Nesting must be tidy:** close children before parents — `<ul><li>…</li></ul>`, never `<ul><li>…</ul></li>`.
> - **Tag names are lowercase** by convention (`<section>`, not `<SECTION>`).

### 📖 Reference & glossary

| Term | Plain meaning |
|---|---|
| **HTML** | the language that describes a web page's structure |
| **Element / tag** | a unit of content: `<tag>content</tag>` |
| **Attribute** | extra info in an opening tag: `href="…"`, `class="…"`, `id="…"` |
| **Nesting** | elements placed inside other elements |
| **DOM tree** | the page seen as a tree of nested elements |
| **Semantic tags** | tags that name a region: `header`, `main`, `section`, `footer` |
| **`table` / `tr` / `th` / `td`** | table / row / header cell / data cell |
| **`ul` / `li`** | unordered (bulleted) list / list item |
| **`a` + `href`** | a link and its destination |
| **Fragment vs document** | a snippet vs a full `<!DOCTYPE html>…</html>` file |
