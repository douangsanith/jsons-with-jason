### ⚙️ Setup

````python
from IPython.display import HTML, display

def show(html: str):
    display(HTML(html))

print("Ready.")
````

### 1 · The self-contained file — one document, everything inline

Saved to disk, a Claude artifact is a full HTML **document**: a `<head>` (metadata the visitor doesn't see) and a `<body>` (the page). Everything it needs — styles, scripts, even images — is **inlined**, so the single file runs anywhere with no dependencies.

````python
skeleton = """<!DOCTYPE html>            <!-- 'this is modern HTML' -->
<html lang="en">
  <head>
    <meta charset="utf-8">           <!-- text encoding -->
    <title>Matter Intelligence</title>  <!-- browser-tab title (not shown on page) -->
    <style> /* all CSS lives here */ </style>
  </head>
  <body>
    <!-- all visible content lives here -->
    <script> /* all JavaScript lives here */ </script>
  </body>
</html>"""
print(skeleton)
````

**What just happened:** that's the container every artifact fills in. Four parts to find every time: **`<!DOCTYPE html>`** (the declaration), **`<head>`** (title + `<style>`), **`<body>`** (content), and often a **`<script>`** near the end of the body.

### 2 · `<head>` vs `<body>`

- **`<head>`** — *metadata*: the `<title>` (tab name), `<meta>` (encoding, mobile viewport), and the `<style>` block. None of it renders on the page directly.
- **`<body>`** — everything the visitor **sees**: your Day 1–3 header/main/cards/footer.

When editing, **brand/layout changes → `<head>` `<style>`; content changes → `<body>`.**

### 3 · The inline `<script>` — and why it's quiet here

Interactive artifacts include a **`<script>`**: JavaScript that runs in the browser — filtering a table, toggling a theme, drawing a chart. Two things to know:
- It's usually near the **end of `<body>`** (so the HTML exists before the script runs).
- **It does *not* execute in notebook output** — Jupyter sandboxes `<script>` for safety. To see it run, **open the saved `.html` file in a browser.**

````python
# This renders visually, but the <script> won't run inside the notebook (by design).
show("""
<div id="greet" style="font-family:sans-serif;padding:10px;border:1px solid #e5e7eb;border-radius:8px;">
  <strong>Matter Intelligence</strong> — <span id="count">3</span> active matters
  <button onclick="alert('Full docket lands in Week 5!')">Details</button>
</div>
<script>
  // In a browser this line runs; in notebook output it's ignored.
  document.getElementById('count').textContent = '3 (live)';
</script>
""")
````

**What just happened:** you see the card and button, but the `<script>` line didn't change the count — the notebook blocked it. That's expected. **JavaScript is Week 4's whole topic;** today you just need to *recognize* the `<script>` block and know it runs in the browser, not here.

### 4 · Self-contained tricks: data URIs & escaping

Two things that make one file stand alone:
- **Data URIs** — an image embedded *as text* right in the HTML: `<img src="data:image/png;base64,iVBOR...">`. No separate image file — the bytes are inline. (That's why Claude artifacts have no missing images.)
- **Escaping** — to show characters that mean something in HTML, you write **entities**: `&amp;` → `&`, `&lt;` → `<`, `&gt;` → `>`. So "M&A" is written `M&amp;A` in the source.

````python
show("<p>Practice area: <strong>M&amp;A</strong> &nbsp; (written <code>M&amp;amp;A</code> in the source)</p>")
````

**What just happened:** the page shows **M&A**, but the source wrote `M&amp;A` — because a bare `&` is reserved. When you edit artifact text and see `&amp;`/`&lt;`, that's escaping, not a typo.

### 5 · The modification playbook

When your team wants to change a Claude artifact, sort the edit into one of three buckets:

| ✅ Safe to change yourself | ⚠️ Change carefully | 🧑‍💻 Hand to a developer |
|---|---|---|
| Text & labels in `<body>` | Colors/spacing in `<style>` | Logic inside `<script>` |
| The data in a table/list | Layout (`flex`/`grid`) values | Anything with `fetch`/APIs |
| Title in `<head>` | Adding a new card/section | Build tooling / new dependencies |

**Rule of thumb:** content and styling are yours; **program logic (the `<script>`) is where Week 4 starts.** Until then, if a change is in the script, note it and pair with a developer.

> **`Go Deeper 🔧` — why the sandbox, and how artifacts get hosted.** Notebooks (and many hosts) block inline `<script>` and outside network calls via a **Content Security Policy** — a safety boundary so a page can't run arbitrary code or phone home. Self-contained artifacts (everything inlined, no external requests) are exactly what those policies allow, which is part of why Claude builds them that way. To *run* the JS, open the file locally or host it somewhere that permits it.

> **`Common pitfalls ⚠️`**
>
> - **Don't delete the `<head>` or `<style>`** trying to "clean up" — you'll strip the styling.
> - **Balance the big containers:** every `<html>`, `<head>`, `<body>`, `<script>` needs its closing tag.
> - **Editing text near `&`/`<`/`>`?** Keep the entity form (`&amp;`) or the page may break.
> - **JS not working in the notebook?** That's expected — open the saved `.html` in a browser.

### 📖 Reference & glossary

| Term | Plain meaning |
|---|---|
| **`<!DOCTYPE html>`** | declares a modern HTML document |
| **`<head>`** | metadata: `<title>`, `<meta>`, `<style>` (not shown on page) |
| **`<body>`** | the visible content |
| **`<script>`** | JavaScript (behavior); runs in a browser |
| **Self-contained** | everything inlined into one file (CSS, JS, assets) |
| **Data URI** | an asset embedded as inline text (`data:image/png;base64,…`) |
| **Escaping / entities** | `&amp;` `&lt;` `&gt;` for literal `&` `<` `>` |
| **CSP / sandbox** | safety policy that blocks inline scripts & external calls |
