### 1 · A variable is a defined term

You create a variable with `name = value`. The `=` means *“store the value on the right under the name on the left.”* It is **not** “equals” in the math sense.

Analogy: `client = "Acme Corp"` is exactly like a contract saying **“Client” means Acme Corp**.

````python
client = "Acme Corp"     # give the value "Acme Corp" the name `client`
matter_number = "M-1001"

print(client)
print(matter_number)
````

**What just happened:** we defined two terms and asked Python to print them back. Change `"Acme Corp"` to your own text and re-run.

### 2 · The four everyday data types

Python cares what *kind* of value you're storing. The four you'll use constantly:

| Type | Python name | Legal example |
|---|---|---|
| Text | `str` (“string”) | client name, matter number |
| Whole number | `int` | number of documents |
| Decimal | `float` | amount billed |
| True/False | `bool` | is this privileged? |

Text goes in quotes. Numbers don't. `True`/`False` are capitalized and have no quotes.

````python
client         = "Acme Corp"   # str  -> text, in quotes
num_documents  = 342            # int  -> whole number
amount_billed  = 18500.00       # float-> has a decimal point
is_privileged  = True           # bool -> True or False (no quotes)

# type() tells you what kind of value something is
print(type(client))
print(type(num_documents))
print(type(amount_billed))
print(type(is_privileged))
````

**What just happened:** `type()` reports the category of each value. Notice `"342"` (quotes) would be *text*, but `342` (no quotes) is a *number* you can do math with.

### 3 · Doing a little math

Because numbers are stored as real numbers, you can compute with them — a first taste of *thinking with data* instead of eyeballing a spreadsheet.

````python
hours      = 12
rate       = 450.0          # dollars per hour
fee        = hours * rate   # multiply

print(fee)                  # 5400.0

# variables can be built from other variables
discount   = 0.10           # 10%
final_fee  = fee * (1 - discount)
print(final_fee)
````

### 4 · f-strings: build a sentence from variables

An **f-string** lets you drop variables straight into text. Put an `f` before the quote, then wrap any variable in `{curly braces}`. This is how every report, summary, and log line gets built.

````python
matter_number = input("Enter matter number: ")
client = input("Enter client name: ")
amount_billed = float(input("Enter amount billed: "))

summary = f"Matter {matter_number} for {client}: ${amount_billed:,.2f} billed."
print(summary)

# {amount_billed:,.2f} means: comma thousands + exactly 2 decimals -> 18,500.00
````

> **`Go Deeper 🔧` — converting between types.** Sometimes data arrives as text (“342”) but you need a number. Use `int()`, `float()`, and `str()` to convert. And beware: `"5" + 5` is an error (text + number), but `int("5") + 5` gives `10`.

````python
# Go Deeper (optional) — type conversion
text_count = "342"                # this is TEXT
real_count = int(text_count)      # now it's a number
print(real_count + 8)             # 350

# going the other way, str() turns a number into text for display
print("Docs: " + str(real_count))
````

> **`Common pitfalls ⚠️`**
>
> - Text needs **quotes** (`"M-1001"`); forgetting them makes Python think it's a variable name.
> - `=` **assigns** a value; `==` **compares** two values (you'll use `==` on Day 2).
> - `"5" + 5` → error. Mixing text and numbers needs a conversion first.

### 🚀 Build the artifact — a matter summary line

Assemble everything into the deliverable: a clean, reusable summary built entirely from variables. This exact pattern reappears when we later have Claude *generate* these summaries for us.

````python
# --- Matter summary, built from variables ---
matter_number = "M-1001"
client        = "Acme Corp"
practice_area = "Contracts"
amount_billed = 18500.00
is_privileged = True

summary = (
    f"[{matter_number}] {client} — {practice_area}\n"
    f"  Billed to date: ${amount_billed:,.2f}\n"
    f"  Privileged: {is_privileged}"
)
print(summary)
print("\n✅ Shipped: a matter summary generated from variables.")
````
