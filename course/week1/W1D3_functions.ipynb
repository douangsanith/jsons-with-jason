### 1 · Defining and calling a function

`def` **defines** a function (writes the template); calling it with `name()` **runs** it. The indented lines under `def` are the body — what the function does.

````python
def greet_team():
    print("Good morning, legal innovation team!")

# defining it doesn't run it. Calling it does:
greet_team()
greet_team()   # reuse costs nothing
````

**What just happened:** we wrote the instructions once and ran them twice. That reuse is the whole point.

### 2 · Parameters and return values

A useful function takes **input** (parameters) and hands back **output** (`return`). Analogy: a form template with blanks — you supply *client* and *amount*, it returns the finished line. `return` gives you a value you can store in a variable, unlike `print` which only shows it.

````python
def format_matter(client, amount):
    return f"{client}: ${amount:,.2f} billed"

line = format_matter("Acme Corp", 18500.0)   # supply the blanks
print(line)
print(format_matter("Brightline LLC", 42750.5))
````

**What just happened:** `format_matter` took two inputs and *returned* a finished string. We stored one result in `line` and printed another directly.

> **`print` vs `return` — the #1 beginner mix-up.** `print` just displays something on screen. `return` hands a value back to your program so you can keep using it. Tools need `return`.

### 3 · Default values

Give a parameter a default so callers can skip it. Here, `threshold` defaults to 25000 — but any caller can override it for a specific practice group.

````python
def is_high_value(amount, threshold=25000):
    return amount >= threshold        # returns True or False

print(is_high_value(18500))            # uses default 25000 -> False
print(is_high_value(18500, 10000))     # override threshold  -> True
````

### 4 · Functions + loops = a tool

The payoff: combine today's functions with yesterday's loop. Notice how readable the loop becomes when the logic has good names — that readability *is* good engineering.

````python
docket = [
    {"client": "Acme Corp",     "amount": 18500.00},
    {"client": "Foster & Sons",  "amount": 58900.75},
    {"client": "Evergreen Inc",  "amount":  2450.00},
]

def review(matter):
    tag = "HIGH VALUE" if is_high_value(matter["amount"]) else "routine"
    return f'{format_matter(matter["client"], matter["amount"])}  ->  {tag}'

for m in docket:
    print(review(m))
````

> **`Go Deeper 🔧` — docstrings, multiple returns, type hints.** A **docstring** (triple-quoted line under `def`) documents the function. You can return **several values** at once. And **type hints** signal expected types — great for tools others will use.

````python
# Go Deeper (optional)
def summarize(amount: float, threshold: float = 25000) -> tuple:
    """Return (label, gap_to_threshold) for a billed amount."""
    label = "high" if amount >= threshold else "normal"
    gap = threshold - amount
    return label, gap                 # two values -> a tuple

label, gap = summarize(18500)         # unpack both
print(label, gap)
print(summarize.__doc__)              # the docstring is readable at runtime
````

> **`Common pitfalls ⚠️`**
>
> - Forgetting `return` — the function then hands back `None`. If your result is mysteriously `None`, check for a missing `return`.
> - Confusing `print` (shows) with `return` (hands back a usable value).
> - Indentation matters: the function body must be indented consistently (4 spaces is standard).

### 🚀 Build the artifact — a reusable matter-helpers module

Package the lesson into a tidy set of helpers. This is a real deliverable: clean, named functions you can copy into any future notebook — the first bricks of *Matter Intelligence*.

````python
# ---- matter_helpers: reusable functions for the whole course ----
def format_currency(amount):
    """Money as a clean string: 18500.0 -> '$18,500.00'."""
    return f"${amount:,.2f}"

def is_high_value(amount, threshold=25000):
    """True if a billed amount clears the review threshold."""
    return amount >= threshold

def matter_summary(matter):
    """One-line summary of a matter dict, with a review flag."""
    tag = "needs review" if is_high_value(matter["amount"]) else "routine"
    return f'{matter["matter_id"]} | {matter["client"]:<16} | {format_currency(matter["amount"]):>12} | {tag}'

# demo
docket = [
    {"matter_id": "M-1001", "client": "Acme Corp",    "amount": 18500.00},
    {"matter_id": "M-1006", "client": "Foster & Sons", "amount": 58900.75},
]
for m in docket:
    print(matter_summary(m))
print("\n✅ Shipped: a reusable matter-helpers module.")
````
